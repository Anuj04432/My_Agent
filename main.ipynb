```python
from langchain_core.tools import tool
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from dotenv import load_dotenv
import os


load_dotenv()
```

```python
import requests

@tool
def get_weather(city:str):
    """Whats the weather in city"""
    api_key = os.getenv("WEATHER_API_KEY")
    url = "http://api.weatherapi.com/v1/current.json"
    params ={
        "key":api_key,
        "q":city
    }

    data = requests.get(url,params=params)
    response =  data.json()

    return response

```

```python
llm = ChatOllama(model="qwen2.5:3b-instruct")

agent = create_agent(model=llm,
                     tools=[get_weather],
                     system_prompt="Give only the temparature and city name with condition no wxtra details"
                     )
```

```python
response = agent.invoke({'messages':{
    "role":"user",
    "content":"what is the weather in Mumbai(kurla) and in bangalore compare both "
}})


```

```python
print(response["messages"][-1].content)
```

```python
from tavily import TavilyClient
@tool
def web_search(query: str):
    """Search the web for information"""
    api_key = os.getenv("WEB_SEARCH_API_KRY")

    client = TavilyClient(api_key=api_key)

    response = client.search(
        query=query,
        search_depth="advanced"
    )

    return  response


```

```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

llm = ChatOllama(model="qwen2.5:3b-instruct")

agent = create_agent(model=llm,
                     tools=[get_weather,web_search],
                     system_prompt="Give on temperature if there is weather tool, in web search tool return only a title and content")

response = agent.invoke({"messages":{
    "role":"user",
    "content":"what is the weather in bangalore and what is the current news of india"
}})
```

```python
print(response["messages"][-1].content)
```

```python
from langchain_core.tools import tool
from datetime import datetime

@tool

def time(query:str):
    """Give the date time in proper format"""
    now = datetime.now()
    return f"the current time is {now}"
```

```python
from langchain.agents import create_agent
from langchain_openrouter import ChatOpenRouter
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("OPEN_ROUTER_API_KEY")

llm = ChatOpenRouter(model="openrouter/free",api_key=api_key)

agent = create_agent(model=llm,
                     tools=[get_weather,time,web_search],
                     system_prompt="get the data using tools and represent it with the proper format, don't include everything just use what is need")

response = agent.invoke({"messages":
                         {
                             "role":"user",
                             "content":"What are the current ai news list them in the proper form like title nd description with date"
                         }})
```

```python
print(response["messages"][-1].content)
```
